```python
from autogluon.tabular import TabularDataset, TabularPredictor

# Load data
train_data = TabularDataset('train.csv')
test_data = TabularDataset('test.csv')

# Drop irrelevant columns
train_data = train_data.drop(columns=['Name', 'Ticket', 'Cabin'])

# Train
predictor = TabularPredictor(label='Survived', eval_metric='accuracy') \
    .fit(train_data, time_limit=500, presets='best_quality')

# Evaluate
results = predictor.fit_summary()

# Predict
predictions = predictor.predict(test_data)
print(predictions.head())
```

```python
lb = predictor.leaderboard(train_data, silent=True)
lb.to_csv("full_leaderboard.csv", index=False)
```
